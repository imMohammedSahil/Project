# 🔬 Research Scope AI
### AI-Based Research Scope Generator & Evidence-Backed Research Landscape Explorer

> **Turn a raw research idea into a real, evidence-backed research landscape — not a list of AI-generated guesses.**

---

### 🌟 System Architecture & Pipeline:
```
Idea Intake (Free-form) 
  ⬇️ 
AI Interpretation & Academic Query Builder (Groq LLM) 
  ⬇️ 
Academic Retrieval Engine (Semantic Scholar & OpenAlex APIs)
  ⬇️ 
Semantic Similarity & Graph Engine (Cosine Similarity + NetworkX)
  ⬇️ 
Evidence-Based Saturation Analysis & 2D Landscape Matrix (Plotly)
  ⬇️ 
Automated Research Scope Generator & Markdown Exporter
```

**Key Highlights for Faculty Demo:**
- 🟢 **Zero Hallucination:** Papers, authors, citations, abstracts, and DOIs are retrieved in real-time from academic repositories.
- 🟣 **Evidence Badges:** `[From Paper]` vs `[AI-Inferred]` literature gap classification.
- 📊 **2D Saturation Matrix:** Interactive scatter plot showing Scope Potential vs Research Saturation.
- 🌐 **Gradio Blocks Interface:** Running live with public `share=True` link.
